<h2 style="color:#48C9B0;">2. Chargement et table canonique d’audit</h2>

</div>

<p>
  Les corpus sélectionnés présentent des formats hétérogènes : QCM à réponse unique ou multiple, questions-réponses ouvertes, cas cliniques et paires de préférences DPO.

Les données sont ensuite projetées dans une <strong>table canonique d’audit</strong> qui rassemble, sous des champs communs, les informations nécessaires aux analyses suivantes : source, split, type de tâche, question, contexte, réponse attendue, réponses préférée et rejetée, métadonnées et identifiants de traçabilité.
  
  Cette représentation constitue une couche d’observation commune et ne remplace pas les données sources : les DataFrames originaux restent conservés afin de pouvoir contrôler à tout moment les schémas, les types et les valeurs tels qu’ils existent dans les fichiers bruts.
</p>

```python
# Chargement

# Charge CSV, JSON ou JSONL selon l'extension.
# Le JSON est relu en JSONL si nécessaire.
def load_dataset(path: Path) -> pd.DataFrame:
    """Charge un dataset tabulaire."""
    if path.suffix.lower() == ".csv":
        return pd.read_csv(path)

    if path.suffix.lower() in {".json", ".jsonl"}:
        try:
            return pd.read_json(path)
        except ValueError:
            return pd.read_json(path, lines=True)

    raise ValueError(f"Format non géré : {path.suffix}")


# Validation

# Vérifie les colonnes requises selon la famille du dataset.
# L'identifiant source est ajouté au contrôle lorsqu'il existe.
def validate_schema(name: str, df: pd.DataFrame) -> None:
    """Valide le schéma source."""
    spec = DATASET_SPECS[name]
    required = set(REQUIRED_COLUMNS[spec["family"]])

    if spec["id_col"]:
        required.add(spec["id_col"])

    missing = sorted(required - set(df.columns))

    if missing:
        raise ValueError(f"{name} : colonnes absentes : {missing}")


# Fichiers sources

# Résout les chemins et bloque immédiatement si un fichier manque.
# Aucun dataset n'est chargé avant validation de leur présence.
paths = {
    name: RAW_DIR / spec["path"]
    for name, spec in DATASET_SPECS.items()
}

missing = [str(path) for path in paths.values() if not path.exists()]

if missing:
    raise FileNotFoundError("Fichiers manquants :\n- " + "\n- ".join(missing))


# Chargement et contrôle

# Charge tous les datasets puis valide leur schéma source.
# Les volumes bruts sont conservés pour la traçabilité.
datasets = {
    name: load_dataset(path)
    for name, path in paths.items()
}

for name, df in datasets.items():
    validate_schema(name, df)

raw_row_counts = {
    name: len(df)
    for name, df in datasets.items()
}

display(
    pd.DataFrame(
        raw_row_counts.items(),
        columns=["dataset", "rows"]
    )
)
```

<h4 style="color:#48C9B0;">Synthèse</h4>

<div style="border-left:4px solid #48C9B0; padding-left:15px; margin-left:5px; margin-bottom:10px;">

  Le chargement confirme la disponibilité de l’ensemble des fichiers attendus. 
  
  Les volumes sont très hétérogènes selon les sources : <strong>UltraMedical-Preference</strong> constitue de loin le corpus le plus volumineux, tandis que <strong>FrenchMedMCQA</strong> est plus restreint. <strong>MediQAl</strong> est réparti entre plusieurs formats et splits, et <strong>MedQuAD</strong> fournit un corpus intermédiaire de <strong>16 407 exemples</strong>. 
  
  Cette diversité de volumes devra être prise en compte lors de la sélection des données afin d’éviter qu’une source ne domine excessivement les datasets finaux.
</div>

```python
# Valeurs et texte

# Détecte les valeurs absentes puis convertit proprement toute valeur en texte.
# Les structures sont sérialisées de façon stable pour garantir la reproductibilité.
def is_missing(value) -> bool:
    return (
        value is None
        or isinstance(value, float) and pd.isna(value)
        or isinstance(value, str) and value.strip().lower() in {"", "nan", "none", "null"}
    )


def text_value(value) -> str:
    if is_missing(value):
        return ""
    if isinstance(value, (dict, list, tuple)):
        return json.dumps(value, ensure_ascii=False, sort_keys=True, default=str)
    return str(value).strip()


# Audit des valeurs requises

# Compte les valeurs absentes dans les colonnes obligatoires de chaque famille.
# Le taux permet de mesurer rapidement la complétude des sources.
def required_value_audit(datasets: dict[str, pd.DataFrame]) -> pd.DataFrame:
    """Audite les valeurs requises."""
    rows = []

    for name, df in datasets.items():
        required = REQUIRED_COLUMNS[DATASET_SPECS[name]["family"]]

        for column in sorted(required):
            missing = int(df[column].map(is_missing).sum())
            rows.append({
                "dataset": name,
                "column": column,
                "missing": missing,
                "missing_rate_pct": round(missing / len(df) * 100, 2)
            })

    return pd.DataFrame(rows)


# Normalisation et hash

# Normalise le texte pour les comparaisons et génère des identifiants reproductibles.
# Les accents, espaces et éventuellement la ponctuation sont neutralisés.
def normalize_text(value, *, remove_punctuation: bool = False) -> str:
    text = unicodedata.normalize("NFKD", text_value(value).lower())
    text = "".join(c for c in text if not unicodedata.combining(c))

    if remove_punctuation:
        text = re.sub(r"[^\w\s]", " ", text)

    return re.sub(r"\s+", " ", text).strip()


def stable_hash(*values) -> str:
    payload = "\u241f".join(normalize_text(value) for value in values)
    return hashlib.sha256(payload.encode()).hexdigest()


# Réponses QCM

# Extrait les positions A-E et harmonise les choix FrenchMedMCQA / MediQAl.
# Les réponses invalides ou absentes sont ignorées.
def extract_positions(value) -> list[str]:
    if is_missing(value):
        return []

    if isinstance(value, str):
        try:
            value = ast.literal_eval(value)
        except (ValueError, SyntaxError):
            pass

    items = value if isinstance(value, (list, tuple, set)) else [value]
    positions = []

    for item in items:
        compact = re.sub(r"[^A-E]", "", str(item).upper())
        if 0 < len(compact) <= 5:
            positions.extend(compact)

    return list(dict.fromkeys(positions))


def fmm_choices(row: pd.Series) -> dict[str, str]:
    value = row.get("answers")

    if isinstance(value, dict):
        choices = {str(k).upper(): text_value(v) for k, v in value.items()}
        return {key: choices[key] for key in "ABCDE" if choices.get(key)}

    if isinstance(value, (list, tuple)):
        return {key: text_value(value) for key, value in zip("ABCDE", value)}

    return {}


def mediqal_choices(row: pd.Series) -> dict[str, str]:
    return {
        key.upper(): text_value(row.get(f"answer_{key}"))
        for key in "abcde"
        if text_value(row.get(f"answer_{key}"))
    }


# Réponse assistant

# Extrait la dernière réponse assistant depuis les formats texte, dictionnaire ou chat.
# Les autres formats sont convertis proprement en texte.
def extract_assistant_text(value) -> str:
    if isinstance(value, str):
        return value.strip()

    if isinstance(value, dict):
        return text_value(value.get("content", value))

    if isinstance(value, list):
        answers = [
            text_value(item.get("content"))
            for item in value
            if isinstance(item, dict) and item.get("role") == "assistant"
        ]
        if answers:
            return answers[-1]

    return text_value(value)


# Construction QCM

# Assemble contexte, question et choix dans un texte homogène.
# Le contexte est omis lorsqu'il est absent.
def render_qcm(context: str, question: str, choices: dict[str, str]) -> str:
    parts = [context, "", question] if context else [question]
    parts.extend(f"{key}. {value}" for key, value in choices.items())
    return "\n".join(parts).strip()
```

```python
# Canonisation

# Convertit chaque format source vers un schéma canonique commun.
# Les règles propres à chaque famille restent explicites et auditables.
def canonical_record(dataset_name: str, row_index, row: pd.Series) -> dict:
    spec = DATASET_SPECS[dataset_name]
    source, family = spec["source"], spec["family"]

    source_id = (
        text_value(row.get(spec["id_col"]))
        if spec["id_col"]
        else f"{dataset_name}:{row_index}"
    )

    question = context = target = chosen = rejected = ""
    subject = question_type = label_type = ""
    choices, positions = {}, []
    declared_correct_count = None
    metadata = row.get("metadata") if isinstance(row.get("metadata"), dict) else {}

    if family == "qcm_fmm":
        question = text_value(row.get("question"))
        choices = fmm_choices(row)
        positions = extract_positions(row.get("correct_answers"))
        target = ",".join(positions)
        subject = text_value(row.get("subject_name"))
        question_type = text_value(row.get("type"))
        declared_correct_count = row.get("nbr_correct_answers")
        input_text = render_qcm("", question, choices)
        group_text, task_type = input_text, "QCM"

    elif family in {"qcm_mcqm", "qcm_mcqu"}:
        context = text_value(row.get("clinical_case"))
        question = text_value(row.get("question"))
        choices = mediqal_choices(row)
        positions = extract_positions(row.get("correct_answers"))
        target = ",".join(positions)
        subject = text_value(row.get("medical_subject"))
        question_type = text_value(row.get("question_type"))
        input_text = render_qcm(context, question, choices)
        group_text = context or input_text
        task_type = "MCQM" if family == "qcm_mcqm" else "MCQU"

    elif family == "open_oeq":
        context = text_value(row.get("clinical_case"))
        question = text_value(row.get("question"))
        target = text_value(row.get("answer"))
        subject = text_value(row.get("medical_subject"))
        question_type = text_value(row.get("question_type"))
        input_text = "\n\n".join(filter(None, [context, question]))
        group_text, task_type = context or question, "OPEN_QA"

    elif family == "open_qa":
        question = text_value(row.get("Question"))
        target = text_value(row.get("Answer"))
        question_type = text_value(row.get("qtype"))
        input_text = group_text = question
        task_type = "OPEN_QA"

    elif family == "dpo":
        question = text_value(row.get("prompt"))
        chosen = extract_assistant_text(row.get("chosen"))
        rejected = extract_assistant_text(row.get("rejected"))
        label_type = text_value(row.get("label_type"))
        input_text = group_text = question
        task_type = "DPO"

    else:
        raise ValueError(f"Famille non gérée : {family}")

    # Enregistrement canonique

    # Construit le texte complet et un identifiant stable pour chaque source.
    # Le schéma commun conserve les informations nécessaires aux audits suivants.
    full_text = "\n\n".join(
        filter(None, [input_text, target, chosen, rejected])
    )

    return {
        "id": stable_hash(dataset_name, row_index, source_id)[:24],
        "dataset": dataset_name,
        "source": source,
        "source_id": source_id,
        "source_index": row_index,
        "split": spec["split"],
        "family": family,
        "usage": spec["usage"],
        "expected_language": SOURCE_REGISTRY[source]["language"],
        "allowed_for_training": SOURCE_REGISTRY[source]["allowed_for_training"],
        "task_type": task_type,
        "question_text": question,
        "context_text": context,
        "input_text": input_text,
        "target_text": target,
        "chosen_text": chosen,
        "rejected_text": rejected,
        "full_text": full_text,
        "group_text": group_text,
        "choices": choices,
        "correct_positions": positions,
        "declared_correct_count": declared_correct_count,
        "medical_subject": subject,
        "question_type": question_type,
        "label_type": label_type,
        "metadata": metadata,
        "has_context": bool(context)
    }


# Table canonique

# Canonise tous les datasets dans une table d'audit unique.
# L'unicité des identifiants garantit une ligne traçable par source.
audit_records = pd.DataFrame([
    canonical_record(name, index, row)
    for name, df in datasets.items()
    for index, row in df.iterrows()
])

if not audit_records["id"].is_unique:
    raise ValueError("Identifiants canoniques dupliqués.")

display(
    audit_records
    .groupby(["source", "split", "task_type"])
    .size()
    .rename("rows")
    .reset_index()
)
```

  La table canonique restitue correctement la répartition des données par <strong>source</strong>, <strong>split</strong> et <strong>type de tâche</strong>.
  
  <strong>FrenchMedMCQA</strong> est exclusivement constitué de QCM, <strong>MediQAl</strong> combine MCQU, MCQM et questions ouvertes, <strong>MedQuad-KV</strong> apporte des questions-réponses ouvertes sans split d’origine, tandis qu’<strong>UltraMedical-Preference</strong> fournit les données DPO.

<h2 style="color:#48C9B0;">3. Audit structurel</h2>


<div style="border-left:5px solid #48C9B0; background-color:#f2fbf9; color:#000000; padding:12px 16px; margin:16px 0;">
<strong>Objectif :</strong><br/>Contrôler les schémas, les valeurs manquantes, les identifiants, les différences entre splits et les types internes.
</div>


<p>L’audit original montre que les données sont techniquement exploitables. Les valeurs manquantes sont concentrées dans MediQAl, surtout <code>clinical_case</code> avec environ 54 % à 63 % d’absence dans les QCM. Cette absence peut être structurelle et ne justifie aucune suppression automatique.</p>

```python
# Valeurs essentielles

# Mesure les valeurs absentes dans les champs indispensables de chaque famille.
# Le rapport distingue ainsi les manques critiques des colonnes simplement optionnelles.
def required_value_audit(datasets: dict[str, pd.DataFrame]) -> pd.DataFrame:
    """Audite les valeurs essentielles."""
    rows = []

    for name, df in datasets.items():
        family = DATASET_SPECS[name]["family"]

        for column in sorted(REQUIRED_VALUE_COLUMNS[family]):
            missing = int(df[column].map(is_missing).sum())
            rows.append({
                "dataset": name,
                "column": column,
                "missing": missing,
                "missing_rate_pct": round(missing / len(df) * 100, 2)
            })

    return pd.DataFrame(rows)


# Audit structurel

# Contrôle schémas, valeurs manquantes, types internes et identifiants sources.
# Compare également les colonnes des différents splits d'une même source.
def structural_audit(datasets: dict[str, pd.DataFrame]) -> dict[str, pd.DataFrame]:
    """Audite la structure des datasets."""
    schema, missing, identifiers, types = [], [], [], []

    for name, df in datasets.items():
        spec = DATASET_SPECS[name]

        schema.append({
            "dataset": name,
            "source": spec["source"],
            "split": spec["split"],
            "rows": len(df),
            "columns": len(df.columns),
            "column_names": list(df.columns)
        })

        for column in df.columns:
            count = int(df[column].map(is_missing).sum())

            if count:
                missing.append({
                    "dataset": name,
                    "column": column,
                    "missing": count,
                    "missing_rate_pct": round(count / len(df) * 100, 2)
                })

            type_counts = (
                df[column]
                .dropna()
                .map(lambda value: type(value).__name__)
                .value_counts()
                .to_dict()
            )

            types.append({
                "dataset": name,
                "column": column,
                "pandas_dtype": str(df[column].dtype),
                "python_types": type_counts
            })

        id_col = spec["id_col"]

        if id_col and id_col in df.columns:
            missing_mask = df[id_col].map(is_missing)
            values = df.loc[~missing_mask, id_col]

            identifiers.append({
                "dataset": name,
                "id_column": id_col,
                "missing_ids": int(missing_mask.sum()),
                "unique_ids": int(values.nunique()),
                "repeated_id_rows": int(values.duplicated(keep=False).sum())
            })
        else:
            identifiers.append({
                "dataset": name,
                "id_column": None,
                "missing_ids": None,
                "unique_ids": None,
                "repeated_id_rows": None
            })

    # Comparaison des splits

    # Compare deux à deux les schémas des datasets provenant d'une même source.
    # Les écarts de colonnes sont conservés explicitement dans le rapport.
    splits = []

    for source in sorted({spec["source"] for spec in DATASET_SPECS.values()}):
        names = [
            name for name, spec in DATASET_SPECS.items()
            if spec["source"] == source
        ]

        for index, left in enumerate(names):
            for right in names[index + 1:]:
                left_cols = set(datasets[left].columns)
                right_cols = set(datasets[right].columns)

                splits.append({
                    "source": source,
                    "left_dataset": left,
                    "right_dataset": right,
                    "only_left": sorted(left_cols - right_cols),
                    "only_right": sorted(right_cols - left_cols),
                    "common_columns": len(left_cols & right_cols)
                })

    return {
        "03_schema": pd.DataFrame(schema),
        "03_missing_by_column": pd.DataFrame(missing),
        "03_identifiers": pd.DataFrame(identifiers),
        "03_split_schema_differences": pd.DataFrame(splits),
        "03_internal_types": pd.DataFrame(types)
    }


# Exécution

# Exécute les audits structurels et ajoute le contrôle des valeurs essentielles.
# Les rapports sont centralisés pour leur affichage et leur export ultérieur.
REPORTS = structural_audit(datasets)
REPORTS["03_required_values"] = required_value_audit(datasets)

for name in [
    "03_schema",
    "03_missing_by_column",
    "03_required_values",
    "03_identifiers",
    "03_split_schema_differences"
]:
    print(name)
    display(REPORTS[name])
```

<div style="border-left:4px solid #48C9B0; padding-left:15px; margin-left:5px; margin-bottom:10px;">

  <p style="margin-top:0;">
    L’audit structurel confirme que les corpus sont globalement cohérents avec leur documentation, mais met en évidence plusieurs particularités : 
  </p>

  <p>
    <strong>Pour la structure des fichiers :</strong> 
    <ul>
      <li>
        les splits <code>train</code>,<code>validation</code> et <code>test</code> de <strong>MediQAl MCQM/MCQU</strong>
        présentent un schéma homogène de 12 colonnes. 
      </li>
      <li>
        Les différences observées avec <strong>MediQAl OEQ</strong> sont normales puisqu’il s’agit d’un autre type de tâche,
        basé sur une réponse ouverte plutôt que sur des choix A-E.
      </li>
      <li>
        <strong>FrenchMedMCQA</strong> présente une différence mineure : la colonne <code>type</code> est absente du split <code>test</code>.
      <li>
      De même, la colonne <code>feedback</code> d’<strong>UltraMedical-Preference</strong>
      est absente du split <code>test</code>.
      </li>
    </ul>
  </p>

  <p>
    <strong>Valeurs manquantes :</strong> elles concernent principalement
    <code>clinical_case</code> dans <strong>MediQAl</strong>. Plus de la moitié des
    exemples MCQM et MCQU ne possèdent pas de cas clinique, avec des taux compris
    entre environ <strong>54 % et 63 %</strong>. Cette absence ne doit pas être
    considérée automatiquement comme une anomalie : certaines questions sont autonomes
    et ne nécessitent pas de contexte clinique. Pour OEQ, environ <strong>20 %</strong>
    des valeurs de <code>clinical_case</code> et <code>cc_question_number</code>

  <p>
    <strong>Identifiants :</strong> MediQAl ne présente aucun identifiant manquant

  <p>
    <strong>UltraMedical-Preference :</strong> le split train contient
    <strong>109 353 lignes pour 77 046 prompt_id uniques</strong>, avec
    <strong>59 229 lignes appartenant à des identifiants répétés</strong>.
    Ces répétitions ne doivent pas être supprimées sur le seul critère du
    <code>prompt_id</code>, puisqu’un même prompt peut être associé à plusieurs couples
    de réponses ou préférences. Leur validité devra être déterminée par l’audit DPO
    des triplets <code>prompt / chosen / rejected</code>.
  </p>

  <p style="margin-bottom:6px;">
    <strong>Décisions pour le nettoyage :</strong>
  </p>

  <ul style="line-height:1.6; margin-top:0; margin-bottom:0;">
    <li>ne pas supprimer automatiquement les exemples MediQAl dont <code>clinical_case</code> est absent ; vérifier d’abord si la question est autonome ;</li>
    <li>examiner les identifiants répétés de <strong>FrenchMedMCQA train</strong> avant toute suppression ;</li>
    <li>générer un identifiant stable pour <strong>MedQuad-KV</strong> afin de garantir la traçabilité ;</li>
    <li>ne pas utiliser les répétitions de <code>prompt_id</code> comme critère de déduplication DPO ; utiliser le contenu du prompt et des réponses ;</li>
    <li>harmoniser les différences de colonnes entre les splits dans la représentation finale sans modifier les fichiers sources ;</li>
    <li>traiter séparément <strong>MediQAl OEQ</strong> des formats MCQU/MCQM en raison de sa structure et de son usage différents.</li>
  </ul>

<h2 style="color:#48C9B0;">4. Audit des QCM</h2>


<div style="border-left:5px solid #48C9B0; background-color:#f2fbf9; color:#000000; padding:12px 16px; margin:16px 0;">
<strong>Objectif :</strong><br/>Vérifier le nombre de propositions, l’encodage des réponses, le mode unique ou multiple, la validité des positions et les biais de réponse.
</div>


<p>FrenchMedMCQA et MediQAl MCQM utilisent cinq propositions. Trois questions MCQU n’en ont que quatre et doivent être conservées sans inventer une cinquième option. Les positions sont structurellement valides ; MediQAl MCQU présente toutefois un déséquilibre plus marqué, avec C surreprésenté dans train et validation.</p>

```python
# Contrôle QCM

# Vérifie le nombre de choix, les positions correctes et le mode MCQU / MCQM.
# Contrôle aussi la cohérence avec le nombre de réponses déclaré par la source.
qcm = audit_records[
    audit_records["family"].isin({"qcm_fmm", "qcm_mcqm", "qcm_mcqu"})
].copy()

qcm["choice_count"] = qcm["choices"].map(len)
qcm["correct_count"] = qcm["correct_positions"].map(len)

qcm["invalid_positions"] = qcm.apply(
    lambda row: [
        pos for pos in row["correct_positions"]
        if pos not in row["choices"]
    ],
    axis=1
)

qcm["valid_answer_positions"] = qcm["invalid_positions"].map(len).eq(0)

qcm["valid_mode"] = qcm.apply(
    lambda row:
        row["correct_count"] == 1 if row["family"] == "qcm_mcqu"
        else row["correct_count"] >= 2 if row["family"] == "qcm_mcqm"
        else row["correct_count"] >= 1,
    axis=1
)

qcm["declared_count_match"] = qcm.apply(
    lambda row:
        True if is_missing(row["declared_correct_count"])
        else int(row["declared_correct_count"]) == row["correct_count"],
    axis=1
)


# Synthèse

# Agrège les anomalies QCM par dataset pour mesurer leur qualité structurelle.
# Les erreurs de position, de mode et de comptage restent distinguées.
qcm_summary = (
    qcm.groupby("dataset")
    .agg(
        rows=("id", "size"),
        min_choices=("choice_count", "min"),
        max_choices=("choice_count", "max"),
        min_correct=("correct_count", "min"),
        max_correct=("correct_count", "max"),
        invalid_position_rows=("valid_answer_positions", lambda x: int((~x).sum())),
        invalid_mode_rows=("valid_mode", lambda x: int((~x).sum())),
        declared_count_mismatches=("declared_count_match", lambda x: int((~x).sum()))
    )
    .reset_index()
)


# Distribution des réponses

# Mesure la fréquence des positions correctes A-E pour détecter un éventuel biais.
# Les taux sont calculés indépendamment pour chaque dataset.
position_distribution = (
    qcm[["dataset", "correct_positions"]]
    .explode("correct_positions")
    .dropna()
    .groupby(["dataset", "correct_positions"])
    .size()
    .rename("count")
    .reset_index()
)

position_distribution["rate_pct"] = (
    position_distribution["count"]
    / position_distribution.groupby("dataset")["count"].transform("sum")
    * 100
).round(2)


# Rapports

# Enregistre les résultats dans le registre commun puis affiche leur synthèse.
# La vue pivot facilite la comparaison des positions A-E entre datasets.
REPORTS["04_qcm_summary"] = qcm_summary
REPORTS["04_qcm_position_distribution"] = position_distribution

display(qcm_summary)

display(
    position_distribution
    .pivot(
        index="dataset",
        columns="correct_positions",
        values="rate_pct"
    )
    .fillna(0)
)
```

<h2 style="color:#48C9B0;">5. Audit textuel</h2>


<div style="border-left:5px solid #48C9B0; background-color:#f2fbf9; color:#000000; padding:12px 16px; margin:16px 0;">
<strong>Objectif :</strong><br/>Contrôler les langues, les longueurs, les tokens, les anomalies de forme et l’autonomie des questions MediQAl.
</div>


<p>Les échantillons de l’audit original sont majoritairement conformes aux langues attendues : environ 97,2 % de français pour FrenchMedMCQA, 97,4 % pour MediQAl, 97,4 % d’anglais pour MedQuAD et 99,6 % pour UltraMedical. Les questions dépendantes MediQAl avec contexte représentent environ 3,75 % à 7,24 % selon les sous-ensembles.</p>

```python
# Détection de langue

# Détecte la langue uniquement sur un texte suffisamment long et exploitable.
# Les erreurs ou l'absence de langdetect sont converties en statut unknown.
def detect_language_safe(text: str) -> str:
    if not LANGDETECT_AVAILABLE or len(text.strip()) < 20:
        return "unknown"

    try:
        return detect(text)
    except LangDetectException:
        return "unknown"


# Audit linguistique

# Échantillonne chaque dataset avec une graine fixe puis mesure les langues détectées.
# La distribution est comparée à la langue attendue déclarée dans le registre source.
language_rows = []

for name, group in audit_records.groupby("dataset", sort=False):
    sample = group.sample(
        min(LANGUAGE_SAMPLE_SIZE, len(group)),
        random_state=SEED
    )

    counts = (
        sample["input_text"]
        .map(detect_language_safe)
        .value_counts(dropna=False)
    )

    language_rows.extend({
        "dataset": name,
        "expected_language": group["expected_language"].iloc[0],
        "detected_language": language,
        "sample_rows": len(sample),
        "count": int(count),
        "rate_pct": round(count / len(sample) * 100, 2)
    } for language, count in counts.items())


# Rapport

# Enregistre la distribution linguistique dans le registre commun des audits.
# Le rapport permet de repérer les langues inattendues ou non détectées.
language_summary = pd.DataFrame(language_rows)
REPORTS["05_language_sample_distribution"] = language_summary

display(language_summary)
```

```python
# Longueur des textes

# Mesure la longueur en caractères et en mots des principaux champs textuels.
# Les champs vides sont exclus pour ne pas fausser les distributions.
text_fields = [
    "input_text",
    "target_text",
    "context_text",
    "chosen_text",
    "rejected_text",
    "full_text"
]

lengths = []

for field in text_fields:
    df = audit_records[["dataset", "id", field]].copy()
    df = df[df[field].astype(str).str.len().gt(0)]

    df["field"] = field
    df["char_length"] = df[field].str.len()
    df["word_length"] = df[field].str.split().str.len()

    lengths.append(
        df.drop(columns=field)
    )


# Synthèse

# Agrège médiane, P95 et maximum pour repérer les textes longs ou atypiques.
# Ces mesures serviront ensuite au contrôle des longueurs de séquence.
text_lengths = pd.concat(lengths, ignore_index=True)

text_length_summary = (
    text_lengths
    .groupby(["dataset", "field"])
    .agg(
        rows=("id", "size"),
        char_median=("char_length", "median"),
        char_p95=("char_length", lambda x: x.quantile(0.95)),
        char_max=("char_length", "max"),
        word_median=("word_length", "median"),
        word_p95=("word_length", lambda x: x.quantile(0.95)),
        word_max=("word_length", "max")
    )
    .round(2)
    .reset_index()
)


# Rapport

# Enregistre les statistiques dans le registre commun des audits.
# Le tableau permet de comparer rapidement les distributions entre sources.
REPORTS["05_text_length_summary"] = text_length_summary

display(text_length_summary)
```

```python
# Longueur en tokens

# Tokenise les textes avec le tokenizer du modèle cible et mesure leur longueur.
# L'audit permet d'anticiper les troncatures avant SFT et DPO.
token_summary = pd.DataFrame()
token_thresholds = pd.DataFrame()

if RUN_TOKEN_AUDIT:
    try:
        from transformers import AutoTokenizer

        tokenizer = AutoTokenizer.from_pretrained(TOKENIZER_MODEL)

        token_details = audit_records[["dataset", "id"]].copy()
        token_details["token_count"] = audit_records["full_text"].map(
            lambda text: len(tokenizer.encode(text, add_special_tokens=True))
        )

        token_summary = (
            token_details
            .groupby("dataset")["token_count"]
            .agg(
                rows="size",
                median="median",
                p95=lambda x: x.quantile(0.95),
                p99=lambda x: x.quantile(0.99),
                maximum="max"
            )
            .round(2)
            .reset_index()
        )


        # Seuils de séquence

        # Compte les textes dépassant 512, 1024 et 2048 tokens par dataset.
        # Les taux permettent d'évaluer directement le risque de troncature.
        thresholds = []

        for name, group in token_details.groupby("dataset"):
            for threshold in [512, 1024, 2048]:
                count = int(group["token_count"].gt(threshold).sum())
                thresholds.append({
                    "dataset": name,
                    "threshold": threshold,
                    "rows_above": count,
                    "rate_pct": round(count / len(group) * 100, 2)
                })

        token_thresholds = pd.DataFrame(thresholds)

    except (ImportError, OSError) as error:
        print(f"Audit des tokens non exécuté : {error}")


# Rapports

# Enregistre les distributions et dépassements dans le registre des audits.
# Les résultats servent ensuite à justifier la longueur maximale retenue.
REPORTS["05_token_length_summary"] = token_summary
REPORTS["05_token_thresholds"] = token_thresholds

display(token_summary)
display(token_thresholds)
```

```python
# Références contextuelles

# Repère les formulations FR / EN qui dépendent potentiellement d'un contexte précédent.
# Elles servent à identifier les questions MediQAl qui ne sont pas autonomes.
CONTEXT_REFERENCE_PATTERN = re.compile(
    r"\b(?:ce patient|cette patiente|ce cas|dans ce cas|ces résultats|ci-dessus|"
    r"this patient|this case|these results|above|précédent(?:e|es|s)?)\b",
    flags=re.IGNORECASE
)


# Anomalies textuelles

# Détecte les anomalies simples pouvant dégrader la qualité des exemples.
# Les signaux restent descriptifs et ne provoquent aucune suppression automatique.
def text_anomaly_flags(text: str) -> list[str]:
    """Détecte les anomalies textuelles."""
    text = text or ""
    stripped = text.strip()

    if not stripped:
        return ["EMPTY_TEXT"]

    flags = []

    if len(stripped) < 5:
        flags.append("VERY_SHORT_TEXT")
    if re.search(r"[ \t]{3,}", text):
        flags.append("EXCESSIVE_SPACES")
    if re.search(r"(.)\1{4,}", text, flags=re.IGNORECASE):
        flags.append("REPEATED_CHARACTER")
    if re.search(r"\b(\w+)\s+\1\b", text, flags=re.IGNORECASE):
        flags.append("REPEATED_WORD")

    symbol_ratio = sum(
        not char.isalnum() and not char.isspace()
        for char in text
    ) / len(text)

    if symbol_ratio > 0.30:
        flags.append("HIGH_SYMBOL_RATIO")

    return flags


# Synthèse des anomalies

# Applique les contrôles à chaque enregistrement puis agrège les signaux par dataset.
# Le détail reste disponible pour retrouver précisément les lignes concernées.
anomaly_details = audit_records[["dataset", "id", "full_text"]].copy()
anomaly_details["anomaly_flags"] = anomaly_details["full_text"].map(
    text_anomaly_flags
)
anomaly_details["has_anomaly"] = anomaly_details["anomaly_flags"].map(bool)

anomaly_summary = (
    anomaly_details
    .explode("anomaly_flags")
    .dropna(subset=["anomaly_flags"])
    .groupby(["dataset", "anomaly_flags"])
    .size()
    .rename("rows")
    .reset_index()
    .rename(columns={"anomaly_flags": "anomaly"})
)


# Autonomie MediQAl

# Recherche les questions faisant référence à un contexte clinique implicite.
# Distingue les références disposant du contexte de celles réellement orphelines.
autonomy = audit_records[
    audit_records["source"].eq("MediQAl")
].copy()

autonomy["context_reference"] = autonomy["question_text"].str.contains(
    CONTEXT_REFERENCE_PATTERN,
    regex=True,
    na=False
)

autonomy["autonomy_status"] = np.select(
    [
        autonomy["context_reference"] & autonomy["has_context"],
        autonomy["context_reference"] & ~autonomy["has_context"]
    ],
    [
        "DEPENDENT_WITH_CONTEXT",
        "DEPENDENT_WITHOUT_CONTEXT"
    ],
    default="AUTONOMOUS_OR_UNDETECTED"
)

autonomy_summary = (
    autonomy
    .groupby(["dataset", "autonomy_status"])
    .size()
    .rename("rows")
    .reset_index()
)

autonomy_summary["rate_pct"] = (
    autonomy_summary["rows"]
    / autonomy_summary.groupby("dataset")["rows"].transform("sum")
    * 100
).round(2)


# Rapports

# Archive les anomalies textuelles et les statuts d'autonomie dans le registre commun.
# Les deux rapports servent ensuite à cibler les exemples nécessitant une revue.
REPORTS["05_text_anomalies_summary"] = anomaly_summary
REPORTS["05_autonomy_summary"] = autonomy_summary

display(anomaly_summary)
display(autonomy_summary)
```

<h2 style="color:#48C9B0;">6. Répétitions, quasi-doublons et contradictions</h2>


<div style="border-left:5px solid #48C9B0; background-color:#f2fbf9; color:#000000; padding:12px 16px; margin:16px 0;">
<strong>Objectif :</strong><br/>Distinguer les répétitions légitimes des exemples réellement équivalents et des conflits de cible.
</div>


<p>L’audit original détecte notamment 2 510 paires de quasi-doublons candidates dans MedQuAD, 640 dans UltraMedical train et 355 dans MediQAl MCQU train. Il identifie également 129 groupes QCM potentiellement contradictoires dans MediQAl, représentant 280 observations.</p>

<p><strong>Règle centrale :</strong> une réponse identique n’est jamais un doublon à elle seule. Des cas cliniques différents peuvent légitimement conduire à la même réponse.</p>

```python
# Empreintes normalisées

# Normalise entrées et sorties puis calcule des empreintes stables.
# Elles permettent de détecter les répétitions sans dépendre de la forme du texte.
audit_records["input_norm"] = audit_records["input_text"].map(
    lambda x: normalize_text(x, remove_punctuation=True)
)

audit_records["target_norm"] = audit_records.apply(
    lambda row: normalize_text(
        row["target_text"]
        or "\n".join([row["chosen_text"], row["rejected_text"]]),
        remove_punctuation=True
    ),
    axis=1
)

audit_records["input_hash"] = audit_records["input_norm"].map(stable_hash)
audit_records["target_hash"] = audit_records["target_norm"].map(stable_hash)

audit_records["example_hash"] = audit_records.apply(
    lambda row: stable_hash(row["input_norm"], row["target_norm"]),
    axis=1
)

audit_records["group_hash"] = audit_records["group_text"].map(
    lambda x: stable_hash(normalize_text(x, remove_punctuation=True))
)


# Groupes répétés

# Compte les empreintes apparaissant plusieurs fois dans chaque dataset.
# Distingue les entrées répétées des exemples complets réellement répétés.
def repeated_group_summary(
    df: pd.DataFrame,
    key: str,
    label: str
) -> pd.DataFrame:
    """Résume les groupes répétés."""
    counts = (
        df.groupby(["dataset", key])
        .size()
        .rename("group_size")
        .reset_index()
    )

    repeated = counts[counts["group_size"].gt(1)]

    return (
        repeated
        .groupby("dataset")
        .agg(
            repeated_groups=(key, "size"),
            repeated_rows=("group_size", "sum")
        )
        .reset_index()
        .assign(repetition_type=label)
    )


# Détection

# Recherche les répétitions d'exemples complets puis celles des seules entrées.
# Les résultats restent descriptifs et ne suppriment aucun enregistrement.
complete_repetitions = repeated_group_summary(
    audit_records,
    "example_hash",
    "NORMALIZED_COMPLETE"
)

input_repetitions = repeated_group_summary(
    audit_records,
    "input_hash",
    "NORMALIZED_INPUT"
)


# Rapports

# Archive les deux niveaux de répétition dans le registre commun des audits.
# Ils serviront ensuite au contrôle des doublons et des fuites entre splits.
REPORTS["06_normalized_complete_repetitions"] = complete_repetitions
REPORTS["06_normalized_input_repetitions"] = input_repetitions

display(complete_repetitions)
display(input_repetitions)
```

```python
# Doublons proches

# Recherche les entrées similaires avec RapidFuzz dans des groupes de longueur comparable.
# Les doublons exacts et les textes trop différents en longueur sont exclus.
def find_near_duplicates(
    df: pd.DataFrame,
    threshold: int = NEAR_DUPLICATE_THRESHOLD,
    window: int = NEAR_DUPLICATE_WINDOW
) -> pd.DataFrame:
    """Détecte les doublons proches."""
    columns = ["dataset", "first_id", "second_id", "score"]

    if not RAPIDFUZZ_AVAILABLE or not RUN_NEAR_DUPLICATES:
        return pd.DataFrame(columns=columns)

    work = df[df["input_norm"].str.len().gt(0)].copy()
    work["length_bucket"] = work["input_norm"].str.len() // 100
    candidates = []

    for (dataset, _), group in work.groupby(
        ["dataset", "length_bucket"],
        sort=False
    ):
        group = group.sort_values("input_norm", kind="stable")
        indices = list(group.index)

        for pos, first in enumerate(indices):
            text_a = group.at[first, "input_norm"]

            for second in indices[pos + 1:pos + 1 + window]:
                text_b = group.at[second, "input_norm"]

                if text_a == text_b:
                    continue

                length_ratio = min(len(text_a), len(text_b)) / max(
                    len(text_a),
                    len(text_b)
                )

                if length_ratio < 0.80:
                    continue

                score = max(
                    fuzz.ratio(text_a, text_b),
                    fuzz.token_set_ratio(text_a, text_b)
                )

                if score >= threshold:
                    candidates.append({
                        "dataset": dataset,
                        "first_id": group.at[first, "id"],
                        "second_id": group.at[second, "id"],
                        "score": round(float(score), 2)
                    })

    return pd.DataFrame(candidates, columns=columns)


# Synthèse

# Exécute la détection puis compte les paires candidates par dataset.
# Ces similarités restent des signaux d'audit et non des suppressions automatiques.
near_duplicates = find_near_duplicates(audit_records)

near_duplicate_summary = (
    near_duplicates
    .groupby("dataset")
    .size()
    .rename("candidate_pairs")
    .reset_index()
    if not near_duplicates.empty
    else pd.DataFrame(columns=["dataset", "candidate_pairs"])
)


# Rapports

# Conserve le détail des paires et leur synthèse dans le registre des audits.
# Le détail permet une revue ultérieure des similarités détectées.
REPORTS["06_near_duplicates_details"] = near_duplicates
REPORTS["06_near_duplicates_summary"] = near_duplicate_summary

display(near_duplicate_summary)
```

```python
# Vocabulaire médical

# Regroupe les termes médicaux généraux FR / EN utilisés pour qualifier le contenu.
# Ce lexique sert de signal documentaire et non de diagnostic clinique.
MEDICAL_TERMS = {
    "symptom", "symptoms", "symptome", "symptomes", "pain", "douleur",
    "fever", "fievre", "dyspnea", "dyspnee", "breath", "respiration",
    "bleeding", "hemorrhage", "hemorragie", "vomiting", "vomissement",
    "nausea", "nausee", "cough", "toux", "fatigue", "weakness",
    "patient", "patiente", "patients", "disease", "maladie", "syndrome",
    "diagnosis", "diagnostic", "treatment", "traitement", "therapy", "therapie",
    "medication", "medicament", "drug", "dose", "dosage", "blood", "sang",
    "heart", "cardiac", "coeur", "cardiaque", "lung", "pulmonary",
    "poumon", "pulmonaire", "renal", "kidney", "rein", "hepatic", "liver",
    "foie", "neurological", "neurologique", "infection", "infectious",
    "infectieux", "cancer", "tumor", "tumeur", "fracture", "trauma",
    "traumatisme", "pregnancy", "grossesse", "surgery", "chirurgie",
    "emergency", "urgence", "clinical", "clinique", "examination", "examen",
    "test", "analyse", "pressure", "tension", "temperature", "pulse", "pouls",
    "oxygen", "oxygene", "saturation"
}


# Marqueurs de triage

# Regroupe les indices cliniques FR / EN pertinents pour l'analyse du triage.
# Les expressions régulières détectent des signaux sans attribuer de priorité.
TRIAGE_MARKERS = {
    "RESPIRATORY_DISTRESS": [
        r"\bdyspn", r"\bdetresse respiratoire\b", r"\bdifficulte respiratoire\b",
        r"\bshortness of breath\b", r"\brespiratory distress\b", r"\bhypoxi",
        r"\bcyan", r"\bsaturation\b", r"\bspo2\b"
    ],
    "CONSCIOUSNESS": [
        r"\bconscience\b", r"\binconscient", r"\bcoma\b", r"\bconfusion\b",
        r"\bsyncope\b", r"\bunconscious\b", r"\bglasgow\b", r"\bgcs\b"
    ],
    "HEMODYNAMIC": [
        r"\bhypotension\b", r"\bhypertension\b", r"\btension arterielle\b",
        r"\bblood pressure\b", r"\btachycard", r"\bbradycard", r"\bpouls\b",
        r"\bpulse\b", r"\bshock\b", r"\bchoc\b"
    ],
    "BLEEDING": [
        r"\bhemorrag", r"\bsaign", r"\bbleeding\b", r"\bhemorrhag",
        r"\bhematem", r"\bmelena\b"
    ],
    "TRAUMA": [
        r"\btrauma", r"\btraumatis", r"\bfracture\b", r"\baccident\b",
        r"\bchute\b", r"\bfall\b", r"\bhead injury\b"
    ],
    "ACUTE_NEUROLOGICAL": [
        r"\bavc\b", r"\bstroke\b", r"\bconvulsion", r"\bseizure",
        r"\bparalys", r"\bdeficit neurolog", r"\bneurological deficit\b"
    ],
    "ACUTE_PAIN": [
        r"\bdouleur\b", r"\bpain\b", r"\bthorac",
        r"\bchest pain\b", r"\babdominal pain\b"
    ],
    "VITAL_SIGNS": [
        r"\btemperature\b", r"\bfievre\b", r"\bfever\b",
        r"\bfrequence cardiaque\b", r"\bheart rate\b",
        r"\bfrequence respiratoire\b", r"\brespiratory rate\b",
        r"\bsaturation\b", r"\bspo2\b"
    ],
    "ACUTE_CONTEXT": [
        r"\burgence", r"\bemergency\b", r"\baigu", r"\bacute\b",
        r"\bsoudain", r"\bsudden\b", r"\bbrutal",
        r"\bdepuis \d+ (minute|minutes|heure|heures|jour|jours)"
    ]
}


# Situations à risque

# Identifie les situations cliniques nécessitant une vigilance particulière.
# Ces marqueurs alimentent l'audit sans constituer seuls une décision de triage.
RISK_MARKERS = {
    "SELF_HARM_OR_SUICIDE": [
        r"\bsuicid", r"\bself[- ]harm\b", r"\bautomutil",
        r"\btentative de suicide\b"
    ],
    "OVERDOSE_OR_POISONING": [
        r"\boverdose\b", r"\bintoxication\b", r"\bpoison", r"\bsurdosage\b"
    ],
    "SEVERE_BLEEDING": [
        r"\bhemorrag", r"\bmassive bleeding\b", r"\bsaignement massif\b"
    ],
    "CARDIAC_EMERGENCY": [
        r"\barret cardiaque\b", r"\bcardiac arrest\b", r"\binfarct",
        r"\bmyocardial infarction\b"
    ],
    "NEUROLOGICAL_EMERGENCY": [
        r"\bavc\b", r"\bstroke\b", r"\bstatus epilepticus\b",
        r"\betat de mal epileptique\b"
    ],
    "SEVERE_RESPIRATORY": [
        r"\bdetresse respiratoire\b", r"\brespiratory distress\b",
        r"\basphyx", r"\brespiratory failure\b",
        r"\binsuffisance respiratoire\b"
    ],
    "PREGNANCY_OBSTETRIC": [
        r"\bgrossesse\b", r"\bpregnancy\b", r"\beclamps",
        r"\bpre[- ]eclamps", r"\bhemorragie du post[- ]partum\b"
    ],
    "PEDIATRIC": [
        r"\bnourrisson\b", r"\binfant\b", r"\bnewborn\b",
        r"\bnouveau[- ]ne\b", r"\benfant\b", r"\bpediatric\b",
        r"\bpediatrique\b"
    ]
}


# Données de santé sensibles

# Repère les catégories médicales particulièrement sensibles présentes dans le corpus.
# Leur présence est documentée séparément du contrôle d'identifiants personnels.
SENSITIVE_HEALTH_MARKERS = {
    "MENTAL_HEALTH": [
        r"\bdepression\b", r"\bdepressif", r"\bpsychiatr", r"\bschizophren",
        r"\bpsychos", r"\bbipolar", r"\banxiet", r"\bsuicid",
        r"\bself[- ]harm\b"
    ],
    "SEXUAL_REPRODUCTIVE_HEALTH": [
        r"\bgrossesse\b", r"\bpregnancy\b", r"\bcontracept", r"\bsexual",
        r"\bsexuel", r"\bmenstru", r"\bfertilit", r"\binfertilit",
        r"\bivg\b", r"\babortion\b"
    ],
    "INFECTIOUS_DISEASE": [
        r"\bhiv\b", r"\bvih\b", r"\baids\b", r"\bsida\b",
        r"\bhepatit", r"\btuberculos"
    ],
    "GENETIC_INFORMATION": [
        r"\bgenetic", r"\bgenetique\b", r"\bmutation\b",
        r"\bchromosom", r"\bheredit"
    ],
    "ONCOLOGY": [
        r"\bcancer\b", r"\btumeur\b", r"\btumor\b", r"\bcarcinom",
        r"\bleukem", r"\blymphom", r"\bmetasta"
    ],
    "SUBSTANCE_USE": [
        r"\baddiction\b", r"\bdependance\b", r"\btoxicoman", r"\balcool",
        r"\balcohol", r"\bcocaine\b", r"\bheroin", r"\bcannabis\b",
        r"\bopioid"
    ]
}
```

```python
# Détection clinique

# Détecte les catégories cliniques et compte les termes médicaux présents.
# Les marqueurs constituent des signaux d'audit et non des décisions cliniques.
def detect_categories(text: str, markers: dict[str, list[str]]) -> list[str]:
    """Détecte les catégories présentes."""
    text = normalize_text(text)

    return [
        category
        for category, patterns in markers.items()
        if any(re.search(pattern, text) for pattern in patterns)
    ]


def medical_marker_count(text: str) -> int:
    """Compte les termes médicaux distincts."""
    tokens = set(re.findall(r"\b[\w'-]+\b", normalize_text(text)))
    return len(tokens & MEDICAL_TERMS)


# Pertinence médicale

# Mesure la densité de marqueurs médicaux et détecte triage, risque et sensibilité.
# Les catégories sont conservées par enregistrement pour permettre leur audit.
medical_details = audit_records[["dataset", "id", "full_text"]].copy()

medical_details["medical_marker_count"] = medical_details["full_text"].map(
    medical_marker_count
)

medical_details["medical_status"] = pd.cut(
    medical_details["medical_marker_count"],
    bins=[-1, 0, 2, np.inf],
    labels=[
        "NO_MARKER_DETECTED",
        "MEDICAL_MARKERS_PRESENT",
        "MEDICAL_MARKERS_STRONG"
    ]
)

for column, markers in {
    "triage_categories": TRIAGE_MARKERS,
    "risk_categories": RISK_MARKERS,
    "sensitive_health_categories": SENSITIVE_HEALTH_MARKERS
}.items():
    medical_details[column] = medical_details["full_text"].map(
        lambda text, m=markers: detect_categories(text, m)
    )

medical_details["triage_marker_count"] = medical_details[
    "triage_categories"
].map(len)

medical_details["risk_flag"] = medical_details[
    "risk_categories"
].map(bool)


# Synthèses

# Agrège la pertinence médicale, les signaux de triage et les contenus à risque.
# Les seuils distinguent absence, présence et forte présence de marqueurs.
medical_summary = (
    medical_details
    .groupby(["dataset", "medical_status"], observed=True)
    .size()
    .rename("rows")
    .reset_index()
)

medical_details["triage_status"] = np.select(
    [
        medical_details["triage_marker_count"].ge(3),
        medical_details["triage_marker_count"].ge(1)
    ],
    [
        "TRIAGE_MARKERS_STRONG",
        "TRIAGE_MARKERS_PRESENT"
    ],
    default="NO_TRIAGE_MARKER_DETECTED"
)

triage_summary = (
    medical_details
    .groupby(["dataset", "triage_status"])
    .size()
    .rename("rows")
    .reset_index()
)

risk_summary = (
    medical_details
    .groupby("dataset")["risk_flag"]
    .agg(rows="size", flagged_rows="sum")
    .reset_index()
)

risk_summary["flagged_rate_pct"] = (
    risk_summary["flagged_rows"] / risk_summary["rows"] * 100
).round(2)


# Rapports

# Archive les synthèses et le détail clinique sans recopier les textes sources.
# Les résultats servent ensuite à la sélection et aux contrôles de couverture.
REPORTS["08_medical_relevance_summary"] = medical_summary
REPORTS["08_triage_relevance_summary"] = triage_summary
REPORTS["08_risk_content_summary"] = risk_summary
REPORTS["08_medical_details"] = medical_details.drop(columns="full_text")

display(medical_summary)
display(triage_summary)
display(risk_summary)
```

<div style="border-left:5px solid #48C9B0; background-color:#f2fbf9; color:#000000; padding:12px 16px; margin:16px 0;">
<strong>Objectif :</strong><br/>Détecter les identifiants personnels potentiels et distinguer ces signaux des informations médicales sensibles nécessaires à l’apprentissage.
</div>


<p>L’audit original ne détecte aucun candidat PII dans FrenchMedMCQA et MediQAl. Il signale 159 observations MedQuAD et quelques observations UltraMedical, dont 161 dans le train. Ces correspondances restent des candidats à vérifier.</p>


```python
# Audit RGPD avec Presidio

from presidio_analyzer import (
    AnalyzerEngine,
    Pattern,
    PatternRecognizer,
    RecognizerResult
)
from presidio_analyzer.nlp_engine import NlpEngineProvider
from presidio_anonymizer import AnonymizerEngine
from presidio_anonymizer.entities import OperatorConfig
from tqdm.auto import tqdm


# Configuration Presidio

# Configure l'analyse bilingue et les catégories PII recherchées.
# Les entités directes déclenchent anonymisation ou revue humaine.
PRESIDIO_SCORE_THRESHOLD = 0.50

PRESIDIO_MODELS = {
    "fr": "fr_core_news_md",
    "en": "en_core_web_md"
}

PRESIDIO_ENTITIES = [
    "PERSON", "LOCATION", "EMAIL_ADDRESS", "PHONE_NUMBER", "IP_ADDRESS",
    "URL", "DATE_TIME", "MEDICAL_LICENSE", "IDENTIFIER", "BIRTH_DATE",
    "DATE_CANDIDATE", "IDENTITY_INTRODUCTION"
]

DIRECT_PII_ENTITIES = {
    "PERSON", "EMAIL_ADDRESS", "PHONE_NUMBER", "IP_ADDRESS", "URL",
    "MEDICAL_LICENSE", "IDENTIFIER", "BIRTH_DATE", "IDENTITY_INTRODUCTION"
}

CUSTOM_PII_PATTERNS = {
    "EMAIL_ADDRESS": (
        0.85,
        [r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b"]
    ),
    "PHONE_NUMBER": (
        0.80,
        [
            r"(?<!\d)(?:\+33|0033)[ .-]?[1-9](?:[ .-]?\d{2}){4}(?!\d)",
            r"(?<!\d)0[1-9](?:[ .-]?\d{2}){4}(?!\d)"
        ]
    ),
    "IP_ADDRESS": (
        0.80,
        [r"\b(?:\d{1,3}\.){3}\d{1,3}\b"]
    ),
    "URL": (
        0.80,
        [
            r"\bhttps?://[^\s]+",
            r"\bwww\.[^\s]+"
        ]
    ),
    "IDENTIFIER": (
        0.75,
        [
            r"\b(?:patient|dossier|identifiant|id|mrn)"
            r"\s*(?:n[o°.]?|number|#|:)?\s*"
            r"[A-Z0-9][A-Z0-9-]{5,}\b",

            r"\b[12]\s?\d{2}(?:\s?\d{2}){5}\s?\d{2}\b"
        ]
    ),
    "BIRTH_DATE": (
        0.75,
        [
            r"\b(?:né|née)\s+le\s+"
            r"\d{1,2}[/-]\d{1,2}[/-]\d{2,4}\b",

            r"\b(?:date de naissance|dob|born on)"
            r"\s*[:\-]?\s*"
            r"\d{1,2}[/-]\d{1,2}[/-]\d{2,4}\b"
        ]
    ),
    "DATE_CANDIDATE": (
        0.55,
        [r"\b\d{1,2}[/-]\d{1,2}[/-]\d{2,4}\b"]
    ),
    "IDENTITY_INTRODUCTION": (
        0.75,
        [
            r"\b(?:je m['’]appelle|mon nom est|my name is)\s+"
            r"[A-ZÀ-ÖØ-Ý][A-Za-zÀ-ÿ'’-]+"
            r"(?:\s+[A-ZÀ-ÖØ-Ý][A-Za-zÀ-ÿ'’-]+){0,2}\b"
        ]
    )
}


# Initialisation Presidio

# Initialise Presidio avec spaCy FR / EN et les recognizers personnalisés.
# Chaque règle est enregistrée dans les deux langues configurées.
def create_presidio_analyzer() -> AnalyzerEngine:
    """Initialise Presidio bilingue."""
    config = {
        "nlp_engine_name": "spacy",
        "models": [
            {"lang_code": lang, "model_name": model}
            for lang, model in PRESIDIO_MODELS.items()
        ]
    }

    provider = NlpEngineProvider(nlp_configuration=config)
    analyzer = AnalyzerEngine(
        nlp_engine=provider.create_engine(),
        supported_languages=list(PRESIDIO_MODELS)
    )

    for lang in PRESIDIO_MODELS:
        for entity, (score, expressions) in CUSTOM_PII_PATTERNS.items():
            patterns = [
                Pattern(
                    name=f"{entity}_{index}",
                    regex=expression,
                    score=score
                )
                for index, expression in enumerate(expressions, 1)
            ]

            analyzer.registry.add_recognizer(
                PatternRecognizer(
                    supported_entity=entity,
                    supported_language=lang,
                    patterns=patterns
                )
            )

    return analyzer


presidio_analyzer = create_presidio_analyzer()
presidio_anonymizer = AnonymizerEngine()

# Détection PII

# Analyse le texte et conserve uniquement positions, catégories et scores.
# Les résultats identiques sont consolidés avec leur meilleur score.
def analyze_pii(text: str, language: str) -> list[dict]:
    """Détecte les PII sans conserver leur valeur."""
    if not text.strip():
        return []

    results = presidio_analyzer.analyze(
        text=text,
        language=language,
        entities=PRESIDIO_ENTITIES,
        score_threshold=PRESIDIO_SCORE_THRESHOLD
    )

    unique = {}

    for result in results:
        key = (result.entity_type, int(result.start), int(result.end))

        if key not in unique or result.score > unique[key].score:
            unique[key] = result

    return [
        {
            "entity_type": result.entity_type,
            "start": int(result.start),
            "end": int(result.end),
            "score": round(float(result.score), 3)
        }
        for result in sorted(
            unique.values(),
            key=lambda x: (x.start, x.end, x.entity_type)
        )
    ]


def get_pii_categories(results: list[dict]) -> list[str]:
    """Retourne les catégories détectées."""
    return sorted({result["entity_type"] for result in results})


def get_privacy_status(categories: list[str]) -> str:
    """Détermine le statut de confidentialité."""
    if not categories:
        return "CLEAR"

    if set(categories) & DIRECT_PII_ENTITIES:
        return "REDACT_OR_REVIEW"

    return "HUMAN_REVIEW"


# Prévisualisation anonymisée

# Reconstruit les résultats Presidio et masque les PII dans une copie du texte.
# L'analyse n'est pas relancée et le texte original reste inchangé.
def mask_pii_preview(text: str, results: list[dict]) -> str:
    """Produit une copie masquée."""
    if not results:
        return text

    analyzer_results = [
        RecognizerResult(
            entity_type=result["entity_type"],
            start=result["start"],
            end=result["end"],
            score=result["score"]
        )
        for result in results
    ]

    operators = {
        result["entity_type"]: OperatorConfig(
            "replace",
            {"new_value": f"[{result['entity_type']}]"}
        )
        for result in results
    }

    return presidio_anonymizer.anonymize(
        text=text,
        analyzer_results=analyzer_results,
        operators=operators
    ).text
    

# Audit Presidio

# Analyse une seule fois chaque texte unique puis réassocie les résultats aux lignes.
# Le hash évite de répéter les traitements NLP coûteux sur des textes identiques.
pii_details = audit_records[
    ["dataset", "id", "expected_language", "full_text"]
].copy()

pii_details["pii_text_hash"] = pii_details.apply(
    lambda row: stable_hash(row["expected_language"], row["full_text"]),
    axis=1
)

unique_texts = (
    pii_details[
        ["pii_text_hash", "expected_language", "full_text"]
    ]
    .drop_duplicates("pii_text_hash")
    .copy()
)

tqdm.pandas(desc="Audit Presidio")

unique_texts["pii_results"] = unique_texts.progress_apply(
    lambda row: analyze_pii(
        row["full_text"],
        row["expected_language"]
    ),
    axis=1
)

result_map = dict(
    zip(unique_texts["pii_text_hash"], unique_texts["pii_results"])
)

pii_details["pii_results"] = pii_details["pii_text_hash"].map(result_map)
pii_details["pii_categories"] = pii_details["pii_results"].map(
    get_pii_categories
)
pii_details["pii_count"] = pii_details["pii_results"].map(len)
pii_details["pii_flag"] = pii_details["pii_count"].gt(0)
pii_details["privacy_status"] = pii_details["pii_categories"].map(
    get_privacy_status
)


# Synthèses PII

# Agrège le volume détecté, les statuts et les catégories par dataset.
# Les exemples affichés sont masqués et le détail technique exclut le texte brut.
pii_summary = (
    pii_details.groupby("dataset")
    .agg(
        rows=("id", "size"),
        flagged_rows=("pii_flag", "sum"),
        detected_entities=("pii_count", "sum")
    )
    .reset_index()
)

pii_summary["flagged_rate_pct"] = (
    pii_summary["flagged_rows"] / pii_summary["rows"] * 100
).round(2)

pii_status_summary = (
    pii_details
    .groupby(["dataset", "privacy_status"])
    .size()
    .rename("rows")
    .reset_index()
)

pii_status_summary["rate_pct"] = (
    pii_status_summary["rows"]
    / pii_status_summary.groupby("dataset")["rows"].transform("sum")
    * 100
).round(2)

pii_category_summary = (
    pii_details[pii_details["pii_flag"]]
    .explode("pii_categories")
    .groupby(["dataset", "pii_categories"])
    .size()
    .rename("rows")
    .reset_index()
)
# Exemples masqués

pii_examples = (
    pii_details[
        pii_details["pii_flag"]
    ]
    .groupby("dataset")
    .head(5)
    .copy()
)

pii_examples["masked_example"] = pii_examples.apply(
    lambda row: mask_pii_preview(
        row["full_text"],
        row["pii_results"]
    ),
    axis=1
)

pii_examples = pii_examples[
    [
        "dataset",
        "id",
        "expected_language",
        "pii_categories",
        "privacy_status",
        "masked_example"
    ]
]


# Détail technique sans texte brut

pii_review_details = pii_details[
    pii_details["pii_flag"]
][
    [
        "dataset",
        "id",
        "expected_language",
        "pii_text_hash",
        "pii_results",
        "pii_categories",
        "pii_count",
        "privacy_status"
    ]
].copy()


# Données de santé sensibles

sensitive_summary = (
    medical_details[
        [
            "dataset",
            "id",
            "sensitive_health_categories"
        ]
    ]
    .explode("sensitive_health_categories")
    .dropna(subset=["sensitive_health_categories"])
    .groupby(
        [
            "dataset",
            "sensitive_health_categories"
        ]
    )
    .size()
    .rename("rows")
    .reset_index()
)


# Enregistrement des rapports

REPORTS["09_pii_summary"] = pii_summary
REPORTS["09_pii_status_summary"] = pii_status_summary
REPORTS["09_pii_category_summary"] = pii_category_summary
REPORTS["09_pii_review_details"] = pii_review_details
REPORTS["09_pii_masked_examples"] = pii_examples
REPORTS["09_sensitive_health_summary"] = sensitive_summary

display(pii_summary)
display(pii_status_summary)
display(pii_category_summary)
display(sensitive_summary)
```

<h2 style="color:#48C9B0;">10. Audit DPO</h2>


<div style="border-left:5px solid #48C9B0; background-color:#f2fbf9; color:#000000; padding:12px 16px; margin:16px 0;">
<strong>Objectif :</strong><br/>Vérifier la structure d’UltraMedical-Preference, les répétitions de prompts et triplets, la cohérence comparative et les métadonnées d’annotation.
</div>


<p>Les champs essentiels sont complets. Dans le train, 54,79 % des lignes sont associées à un prompt répété et 22,75 % appartiennent à un triplet complet répété. Ces deux phénomènes doivent être traités différemment.</p>


<h4 style="color:#48C9B0;">Synthèse</h4>
<div style="border-left:5px solid #48C9B0; background-color:#f2fbf9; color:#000000; padding:12px 16px; margin:16px 0;">
Conserver les prompts répétés lorsqu’ils portent des comparaisons distinctes, dédupliquer les triplets complets, isoler les inversions et ne jamais considérer chosen comme médicalement correcte par définition.
</div>

```python
# Préparation des données DPO

dpo = audit_records[audit_records["family"].eq("dpo")].copy()

dpo["prompt_norm"] = dpo["input_text"].map(lambda value: normalize_text(value, remove_punctuation=True))
dpo["chosen_norm"] = dpo["chosen_text"].map(lambda value: normalize_text(value, remove_punctuation=True))
dpo["rejected_norm"] = dpo["rejected_text"].map(lambda value: normalize_text(value, remove_punctuation=True))

dpo["triplet_hash"] = dpo.apply(
    lambda row: stable_hash(row["prompt_norm"], row["chosen_norm"], row["rejected_norm"]),
    axis=1,
)

dpo["unordered_pair_hash"] = dpo.apply(
    lambda row: stable_hash(
        row["prompt_norm"],
        *sorted([row["chosen_norm"], row["rejected_norm"]]),
    ),
    axis=1,
)

dpo["chosen_equals_rejected"] = dpo["chosen_norm"].eq(dpo["rejected_norm"])
dpo["chosen_length"] = dpo["chosen_text"].str.len()
dpo["rejected_length"] = dpo["rejected_text"].str.len()

dpo["length_ratio"] = (
    dpo[["chosen_length", "rejected_length"]].max(axis=1)
    / dpo[["chosen_length", "rejected_length"]].min(axis=1).clip(lower=1)
)


# Contrôle de la structure des paires DPO

dpo_structure = (
    dpo.groupby("dataset")
    .agg(
        rows=("id", "size"),
        missing_prompt=("input_text", lambda values: int(values.str.len().eq(0).sum())),
        missing_chosen=("chosen_text", lambda values: int(values.str.len().eq(0).sum())),
        missing_rejected=("rejected_text", lambda values: int(values.str.len().eq(0).sum())),
        chosen_equals_rejected=("chosen_equals_rejected", "sum"),
        strong_length_imbalance=("length_ratio", lambda values: int(values.gt(2).sum())),
    )
    .reset_index()
)


# Détection des répétitions DPO

prompt_repetitions = repeated_group_summary(
    dpo.rename(columns={"prompt_norm": "dpo_prompt_norm"}),
    "dpo_prompt_norm",
    "DPO_PROMPT",
)

triplet_repetitions = repeated_group_summary(
    dpo,
    "triplet_hash",
    "DPO_NORMALIZED_TRIPLET"
)


# Détection des inversions chosen / rejected

inversion_groups = (
    dpo.groupby(["dataset", "unordered_pair_hash"])
    .agg(
        rows=("id", "size"),
        directions=("triplet_hash", "nunique"),
        ids=("id", list),
    )
    .reset_index()
)

inversion_groups = inversion_groups[inversion_groups["directions"].gt(1)]


# Détection des conflits de rôle

role_rows = pd.concat([
    dpo[["dataset", "prompt_norm", "id", "chosen_norm"]]
    .rename(columns={"chosen_norm": "response_norm"})
    .assign(role="chosen"),

    dpo[["dataset", "prompt_norm", "id", "rejected_norm"]]
    .rename(columns={"rejected_norm": "response_norm"})
    .assign(role="rejected"),
], ignore_index=True)

role_conflicts = (
    role_rows.groupby(["dataset", "prompt_norm", "response_norm"])
    .agg(
        roles=("role", "nunique"),
        rows=("id", "size"),
        ids=("id", list),
    )
    .reset_index()
)

role_conflicts = role_conflicts[role_conflicts["roles"].gt(1)]


# Distribution des labels et répétition des prompt_id

label_distribution = (
    dpo.groupby(["dataset", "label_type"])
    .size()
    .rename("rows")
    .reset_index()
)

prompt_id_repetitions = pd.DataFrame([
    {
        "dataset": dataset,
        "rows": len(group),
        "unique_prompt_ids": group["source_id"].nunique(),
        "repeated_prompt_id_rows": int(group["source_id"].duplicated(keep=False).sum()),
    }
    for dataset, group in dpo.groupby("dataset")
])


# Enregistrement et affichage des rapports

REPORTS["10_dpo_structure"] = dpo_structure
REPORTS["10_dpo_prompt_repetitions"] = prompt_repetitions
REPORTS["10_dpo_triplet_repetitions"] = triplet_repetitions
REPORTS["10_dpo_inversions"] = inversion_groups
REPORTS["10_dpo_role_conflicts"] = role_conflicts
REPORTS["10_dpo_label_distribution"] = label_distribution
REPORTS["10_dpo_prompt_id_repetitions"] = prompt_id_repetitions

display(dpo_structure)
display(prompt_repetitions)
display(triplet_repetitions)
display(inversion_groups.head(20))
```

```python
# Métadonnées DPO

# Mesure la couverture des clés de métadonnées pour chaque dataset DPO.
# Contrôle séparément la disponibilité de golden_answer.
metadata_rows, golden_rows = [], []

for dataset, group in dpo.groupby("dataset"):
    keys = Counter()
    golden_missing = 0

    for metadata in group["metadata"]:
        if not isinstance(metadata, dict):
            golden_missing += 1
            continue

        keys.update(map(str, metadata))
        golden_missing += is_missing(metadata.get("golden_answer"))

    metadata_rows.extend(
        {
            "dataset": dataset,
            "metadata_key": key,
            "rows_with_key": int(count),
            "coverage_pct": round(count / len(group) * 100, 2)
        }
        for key, count in keys.items()
    )

    golden_rows.append({
        "dataset": dataset,
        "rows": len(group),
        "missing_golden_answer": int(golden_missing),
        "missing_rate_pct": round(golden_missing / len(group) * 100, 2)
    })


# Rapports

# Construit les tableaux d'audit et les ajoute aux rapports DPO.
# golden_answer reste un contrôle de disponibilité et non de qualité clinique.
metadata_summary = pd.DataFrame(metadata_rows)
golden_answer_summary = pd.DataFrame(golden_rows)

REPORTS["10_dpo_metadata_keys"] = metadata_summary
REPORTS["10_dpo_golden_answer"] = golden_answer_summary

display(metadata_summary)
display(golden_answer_summary)
```

    <strong>Décisions pour le nettoyage DPO :</strong>
  </p>

    <li>conserver uniquement les lignes disposant d’un <code>prompt</code>, d’un <code>chosen</code> et d’un <code>rejected</code> valides et distincts ;</li>

    <li>ne pas exclure automatiquement les lignes dont <code>golden_answer</code> est absent, ce champ n’étant pas indispensable au DPO ;</li>
    <li>utiliser le <code>golden_answer</code> lorsqu’il est disponible comme signal complémentaire de contrôle de la cohérence et de la qualité clinique ;</li>

<h2 style="color:#48C9B0;">11. Audit des fuites entre splits</h2>


<div style="border-left:5px solid #48C9B0; background-color:#f2fbf9; color:#000000; padding:12px 16px; margin:16px 0;">
<strong>Objectif :</strong><br/>Mesurer les chevauchements train, validation et test sur le contenu complet et les groupes indivisibles.
</div>


<p>
Les splits fournis par les sources sont analysés afin de détecter les chevauchements de contenu et de groupes logiques entre entraînement, validation et test. Ces contrôles permettent d'identifier les risques de fuite avant la reconstruction des splits définitifs.
</p>

```python
# Regroupement des corpus

# Rattache train, validation et test à leur corpus source commun.
# Ce regroupement permet de rechercher les fuites uniquement entre splits comparables.
CORPUS_GROUP = {
    **{f"fmmcqa_{s}": "fmmcqa" for s in ["train", "validation", "test"]},
    **{f"mqal_mcqm_{s}": "mqal_mcqm" for s in ["train", "validation", "test"]},
    **{f"mqal_mcqu_{s}": "mqal_mcqu" for s in ["train", "validation", "test"]},
    **{f"umed_{s}": "umed" for s in ["train", "validation", "test"]},
    "mqal_oeq_test": "mqal_oeq",
    "mquad": "mquad"
}

audit_records["corpus_group"] = audit_records["dataset"].map(CORPUS_GROUP)


# Chevauchement entre splits

# Compare deux splits à partir d'une clé normalisée et conserve résumé et détails.
# Le taux mesure la part du split droit appartenant à des groupes déjà présents à gauche.
def split_overlap(
    df: pd.DataFrame,
    key: str,
    left_split: str,
    right_split: str,
    overlap_type: str
) -> tuple[pd.DataFrame, pd.DataFrame]:
    summaries, details = [], []

    for corpus_group, corpus in df.groupby("corpus_group", sort=False):
        left = corpus[corpus["split"].eq(left_split)]
        right = corpus[corpus["split"].eq(right_split)]

        if left.empty or right.empty:
            continue

        left_map = left.groupby(key)["id"].apply(list).to_dict()
        right_map = right.groupby(key)["id"].apply(list).to_dict()
        overlaps = set(left_map) & set(right_map)
        right_rows = sum(len(right_map[value]) for value in overlaps)
        comparison = f"{left_split}_{right_split}"

        summaries.append({
            "corpus_group": corpus_group,
            "comparison": comparison,
            "overlap_type": overlap_type,
            "overlapping_groups": len(overlaps),
            "right_rows_in_overlap": right_rows,
            "right_overlap_rate_pct": round(right_rows / len(right) * 100, 2)
        })

        details.extend({
            "corpus_group": corpus_group,
            "comparison": comparison,
            "overlap_type": overlap_type,
            "group_key": value,
            "left_ids": left_map[value],
            "right_ids": right_map[value]
        } for value in overlaps)

    return pd.DataFrame(summaries), pd.DataFrame(details)


# Audit des fuites

# Teste les trois couples de splits sur les entrées normalisées et groupes cliniques.
# Une présence commune signale une fuite potentielle à contrôler avant entraînement.
leakage_summaries, leakage_details = [], []

for left, right in [
    ("train", "validation"),
    ("train", "test"),
    ("validation", "test")
]:
    for key, overlap_type in [
        ("input_hash", "NORMALIZED_INPUT"),
        ("group_hash", "GROUP")
    ]:
        summary, details = split_overlap(
            audit_records,
            key,
            left,
            right,
            overlap_type
        )
        leakage_summaries.append(summary)
        leakage_details.append(details)

leakage_summary = pd.concat(leakage_summaries, ignore_index=True)
leakage_details_df = pd.concat(leakage_details, ignore_index=True)
```

```python
# Empreinte des fichiers

# Calcule le SHA-256 par blocs pour garantir l'intégrité et la traçabilité des fichiers.
# La lecture par blocs évite de charger entièrement les gros datasets en mémoire.
def file_sha256(path: Path) -> str:
    """Calcule l'empreinte SHA-256 d'un fichier."""
    hasher = hashlib.sha256()

    with path.open("rb") as file:
        for chunk in iter(lambda: file.read(1024 * 1024), b""):
            hasher.update(chunk)

    return hasher.hexdigest()
```

```python
# Intégrité avant export

# Vérifie que la canonicalisation conserve les volumes et des identifiants valides.
# Une incohérence bloque immédiatement l'export des rapports d'audit.
if {name: len(df) for name, df in datasets.items()} != raw_row_counts:
    raise ValueError("Les volumes bruts ont changé.")

if not audit_records["id"].is_unique:
    raise ValueError("Des identifiants canoniques sont dupliqués.")

if audit_records["dataset"].isna().any():
    raise ValueError("Des enregistrements n'ont pas de dataset.")


# Sérialisation JSON

# Convertit les types NumPy et pandas en valeurs compatibles avec JSON.
# Les valeurs manquantes sont converties en null.
def json_ready(value):
    if isinstance(value, np.integer):
        return int(value)
    if isinstance(value, np.floating):
        return float(value)
    if isinstance(value, np.bool_):
        return bool(value)
    if isinstance(value, (pd.Timestamp, datetime)):
        return value.isoformat()
    if isinstance(value, np.ndarray):
        return value.tolist()
    if not isinstance(value, (list, dict, tuple)) and pd.isna(value):
        return None

    return value


# Export des rapports

# Sérialise chaque DataFrame d'audit et retourne ses informations de traçabilité.
# Le manifeste conserve le chemin relatif et le nombre de lignes exportées.
def save_dataframe(df: pd.DataFrame, name: str) -> dict:
    path = AUDIT_DIR / f"{name}.json"
    records = [
        {key: json_ready(value) for key, value in row.items()}
        for row in df.to_dict("records")
    ]

    path.write_text(
        json.dumps(records, ensure_ascii=False, indent=2, default=str),
        encoding="utf-8"
    )

    return {
        "name": name,
        "path": str(path.relative_to(PROJECT_ROOT)),
        "rows": len(df)
    }


export_manifest = [
    save_dataframe(df, name)
    for name, df in REPORTS.items()
    if isinstance(df, pd.DataFrame)
]


# Empreintes des sources

# Calcule le SHA-256 de chaque fichier brut réellement audité.
# Le nettoyage pourra vérifier qu'il travaille exactement sur les mêmes sources.
raw_file_hashes = {
    name: file_sha256(RAW_DIR / spec["path"])
    for name, spec in DATASET_SPECS.items()
}


# Manifeste d'audit

# Centralise provenance, configuration, volumes, empreintes et rapports produits.
# Il constitue la trace reproductible de cette exécution de l'audit.
manifest = {
    "generated_at_utc": datetime.now(timezone.utc).isoformat(),
    "seed": SEED,
    "tokenizer_model": TOKENIZER_MODEL,
    "raw_row_counts": raw_row_counts,
    "raw_file_hashes": raw_file_hashes,
    "canonical_rows": len(audit_records),
    "reports": export_manifest,
    "open_decisions": [
        "Confirm MedQuAD licensing before training use."
    ]
}
```
